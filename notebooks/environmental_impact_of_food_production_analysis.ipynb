### <a id='toc1_1_1_'></a>[1.1. Project Background](#toc0_)
The environmental impact of food production has become a critical global issue, with agriculture contributing significantly to greenhouse gas (GHG) emissions, land use change, water consumption, and pollution (Adedibu, 2023; Cimini, 2021). Research highlights that the expansion of agricultural activities has led to severe ecological problems, including soil degradation, biodiversity loss, and water pollution, necessitating sustainable solutions (Adedibu, 2023). Understanding the sustainability of different food products and their supply chains is crucial in addressing climate change, resource depletion, and ecological balance (Clark et al., 2019; Poore & Nemecek, 2018;  Tilman & Clark, 2014).  

This project aims to analyze the environmental footprint of various food products, comparing **animal-based** and **plant-based** foods across **land use, emissions, water consumption, and eutrophication potential**. Studies have shown that plant-based diets generally have lower environmental footprints than animal-based diets due to lower GHG emissions, reduced land use, and lower water consumption (Aleksandrowicz et al., 2016). The evolution of scientific research on food production’s environmental impact has emphasized the growing need for sustainable agricultural practices and informed consumer choices (Cimini, 2021). By leveraging data-driven insights, we can identify the most sustainable food options and inform policies that promote environmentally friendly food choices (Springmann et al., 2016).  

### <a id='toc1_1_2_'></a>[1.2. Project Objectives](#toc0_)
The primary objectives of this project include:  

1. **Quantifying the Environmental Impact of Different Food Products** – Evaluate the greenhouse gas emissions, land use, and water footprint of various food categories.  
2. **Comparing Animal-Based vs. Plant-Based Foods** – Assess which types of food have a larger environmental impact.  
3. **Identifying Key Environmental Trade-Offs** – Determine whether certain foods have a higher water usage but lower GHG emissions and vice versa.  
4. **Understanding the Supply Chain Contributions** – Analyze which stages (e.g., farm, processing, transport) contribute the most to overall emissions.  
5. **Providing Actionable Insights for Stakeholders** – Support consumers, policymakers, and businesses in making sustainable food choices.

```python
# Define category and subcategory mappings
food_categories = {
    "Animal-Based Foods": {
        "Red Meat": ["Beef (beef herd)", "Beef (dairy herd)", "Lamb & Mutton", "Pig Meat"],
        "Poultry": ["Poultry Meat", "Eggs"],
        "Dairy": ["Milk", "Cheese"],
        "Seafood": ["Fish (farmed)", "Shrimps (farmed)"],
    },
    "Plant-Based Foods": {
        "Cereal Grains & Staples": ["Wheat & Rye (Bread)", "Maize (Meal)", "Barley (Beer)", "Oatmeal", "Rice", "Potatoes", "Cassava"],
        "Legumes & Pulses": ["Other Pulses", "Peas"],
        "Nuts & Seeds": ["Nuts", "Groundnuts"],
        "Soy Products": ["Soymilk", "Tofu"],
        "Vegetable Oils": ["Soybean Oil", "Palm Oil", "Sunflower Oil", "Rapeseed Oil", "Olive Oil"],
        "Vegetables": ["Onions & Leeks", "Root Vegetables", "Brassicas", "Other Vegetables"],
        "Fruits": ["Tomatoes", "Citrus Fruit", "Bananas", "Apples", "Berries & Grapes", "Other Fruit"],
        "Beverages & Alcohol": ["Wine"],
        "Stimulants & Confectionery": ["Coffee", "Dark Chocolate"],
        "Sugars": ["Cane Sugar", "Beet Sugar"],
    }
}

# Function to categorize food products
def categorize_food(product):
    for category, subcategories in food_categories.items():
        for subcategory, items in subcategories.items():
            if product in items:
                return category, subcategory
    return None, None  # If product is not found

# Apply function to create new columns
df_eda[['Food_Category', 'Food_Subcategory']] = df_eda['Food_Product'].apply(lambda x: pd.Series(categorize_food(x)))

# Define the new column order
column_order = [
    "Food_Product", "Food_Category", "Food_Subcategory",  
    "Land_use_Change", "Animal_Feed", "Farm", "Processing", "Transport", 
    "Packaging", "Retail", "Total_emissions", 
    "Eutro_1000kcal", "Eutro_kg", "Eutro_100gP", 
    "Water_1000kcal", "Water_100gP", "Water_kg", 
    "GHG_1000kcal", "GHG_100gP", "Land_1000kcal", "Land_kg", "Land_100gP", 
    "SW_Water_kg", "SW_Water_100gP", "SW_Water_1000kcal"
]

# Reorder the DataFrame
df_eda = df_eda[column_order]

# Display the DataFrame
df_eda.head()
```

```python
# Function to plot categorical food-related information  
def plot_cat_features(data, features, title_fontsize=16, label_fontsize=14, tick_labelsize=12, palette='coolwarm'):  
    num_features = len(features)  
    n_rows = num_features  

    # Create a figure with the required number of rows  
    fig, axes = plt.subplots(n_rows, 1, figsize=(13, 14 * n_rows))  

    # Ensure axes is always a list (needed when only one feature is given)  
    if n_rows == 1:  
        axes = [axes]  

    # Loop through each feature to plot  
    for idx, feature in enumerate(features):  
        # Calculate value counts and sort in descending order  
        value_counts = data[feature].value_counts(normalize=True).sort_values(ascending=False)  
        sorted_categories = value_counts.index  
        total = len(data)  

        # Determine whether to use a horizontal or vertical bar plot  
        if len(sorted_categories) > 20:  
            sns.countplot(y=data[feature], palette=palette, order=sorted_categories, ax=axes[idx])  

            # Add percentage labels for horizontal bar chart  
            for container in axes[idx].containers:  
                labels = [f'{h.get_width()/total*100:.2f}%' if h.get_width() > 0 else '' for h in container]  
                axes[idx].bar_label(container, labels=labels, padding=3)  

            axes[idx].set_ylabel(feature.replace('_', ' ').capitalize(), fontsize=label_fontsize)  
            axes[idx].set_xlabel('Count', fontsize=label_fontsize)  
        else:  
            sns.countplot(x=data[feature], palette=palette, order=sorted_categories, ax=axes[idx])  

            # Add percentage labels for vertical bar chart  
            for container in axes[idx].containers:  
                labels = [f'{h.get_height()/total*100:.2f}%' if h.get_height() > 0 else '' for h in container]  
                axes[idx].bar_label(container, labels=labels, padding=3)  

            axes[idx].set_xlabel(feature.replace('_', ' ').capitalize(), fontsize=label_fontsize)  
            axes[idx].set_ylabel('Count', fontsize=label_fontsize)  
            axes[idx].tick_params(axis='x', rotation=45, labelsize=tick_labelsize)  

        # Set title  
        axes[idx].set_title(f'Distribution of {feature.replace("_", " ")} in Food Data', fontsize=title_fontsize)  
        axes[idx].tick_params(axis='y', labelsize=tick_labelsize)  

    # Display plot  
    plt.tight_layout()  
    plt.show()  

# Plot categorical features related to food emissions analysis  
plot_cat_features(  
    data=df_eda,  
    features=cat_features,  
    title_fontsize=16,  
    label_fontsize=14,  
    tick_labelsize=12,  
    palette='coolwarm'  
)  
```

##### <a id='toc1_2_3_2_3_'></a>[2.3.2.3. Numerical Features](#toc0_)

```python
# Function to analyze numerical food-related features  
def analyse_numeric_features(data, features):  

    # Initialize dictionary to store summary statistics  
    summary_stats = {}  

    for feature in features:  
        # Drop missing values  
        feature_data = data[feature].dropna()  

        # Compute summary statistics  
        mean_val = feature_data.mean()  
        median_val = feature_data.median()  

        summary_stats[feature] = {  
            "count": feature_data.count(),  
            "mean": mean_val,  
            "median": median_val,  
            "std": feature_data.std(),  
            "skewness": feature_data.skew(),  
            "kurtosis": feature_data.kurtosis()  
        }  

        # Create figure with subplots for distribution analysis  
        fig, axes = plt.subplots(1, 2, figsize=(16, 6))  

        # Histogram with KDE to examine distribution  
        axes[0].hist(feature_data, bins=50, alpha=0.5, color="violet", edgecolor="black", density=True, label="Histogram")  
        sns.kdeplot(feature_data, color="darkviolet", linewidth=2, label="KDE", ax=axes[0])  

        # Add Mean & Median Lines for central tendency insights  
        axes[0].axvline(mean_val, color="red", linestyle="dashed", linewidth=2, label=f"Mean: {mean_val:.2f}")  
        axes[0].axvline(median_val, color="blue", linestyle="dashed", linewidth=2, label=f"Median: {median_val:.2f}")  

        # Titles & Labels for histogram  
        axes[0].legend()  
        axes[0].set_title(f"Distribution of {feature} (Histogram & KDE)", fontsize=14, fontweight="bold")  
        axes[0].set_xlabel(feature.replace("_", " "), fontsize=12)  
        axes[0].set_ylabel("Density", fontsize=12)  

        # Box Plot to visualize outliers and spread of values  
        sns.boxplot(y=feature_data, color="skyblue", ax=axes[1])  

        # Title & Labels for boxplot  
        axes[1].set_ylabel(feature.replace("_", " "), fontsize=12)  
        axes[1].set_title(f"Box Plot of {feature}", fontsize=14, fontweight="bold")  

        # Adjust layout and show plot  
        plt.tight_layout()  
        plt.show()  

    # Convert dictionary to DataFrame for summary statistics  
    summary_df = pd.DataFrame(summary_stats).T    

    return summary_df  

# Analyze numerical features related to food emissions  
uni_num_stats = analyse_numeric_features(df_eda, num_features)  
uni_num_stats  
```

```python
def compare_numerical_vs_food_category(df, numerical_features):
    # Define the number of rows and columns for subplots
    number_of_columns = 3
    number_of_rows = math.ceil(len(num_features) / number_of_columns)

    # Create the figure and set the size dynamically based on the number of rows
    fig, axes = plt.subplots(number_of_rows, number_of_columns, figsize=(24, 8 * number_of_rows))

    # Flatten the axes array for easy iteration
    axes = axes.flatten()

    # Iterate through each numerical feature and create a box plot
    for i, feature in enumerate(numerical_features):
        ax = axes[i]
        
        # Create a box plot for the numerical feature grouped by 'FOOD_CATEGORY'
        sns.boxplot(data=df, x='Food_Category', y=feature, palette='coolwarm', ax=ax)
        
        # Set plot title and labels
        ax.set_title(f'Distribution of {feature} vs Food Category', fontsize=15)
        ax.set_xlabel('Food Category', fontsize=12)
        ax.set_ylabel(feature, fontsize=12)
    
    # Remove any unused subplots
    for i in range(len(numerical_features), len(axes)):
        fig.delaxes(axes[i])

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Add suptitle 
    plt.suptitle('Distribution of Numerical Features vs Food Category', fontsize=16, fontweight='bold', y=1.02)
    
    # Show the plot
    plt.show()

# Call the function to visualize the distributions
compare_numerical_vs_food_category(df_eda, num_features)
```

#### <a id='toc1_2_3_5_'></a>[2.3.5. Handling Missing Values and Feature Engineering](#toc0_)

- **Initial Cleanup**:
  - Dropped the `'Food_Category_num'` column as it was not required for analysis.

- **Missing Values Overview**:
  - Created a detailed summary table (`df_aqinfo`) displaying the non-null count, missing count, missing percentage, and data type for each variable.
  - Focused on key environmental impact variables:  
    `Water_100gP`, `SW_Water_100gP`, `Land_100gP`, `Eutro_100gP`, `GHG_100gP`, `Water_1000kcal`, and `SW_Water_1000kcal`.

- **Row-wise Missing Value Handling**:
  - Calculated the number and percentage of missing values across descriptive columns for each row.
  - Dropped 8 rows (≥30% missing values) to maintain data quality and consistency.

- **Grouped Median Imputation**:
  - For `Water_100gP` and `SW_Water_100gP`, missing values were imputed using the **median within each `Food_Subcategory`** group.

- **Regression-based Imputation**:
  - Applied regression models in a defined order (`Land_100gP` → `Eutro_100gP` → `GHG_100gP` → `Water_100gP` → `SW_Water_100gP`) using correlated predictors to estimate missing values.
  - Each model was trained on complete cases and applied to predict missing values for the respective target feature.

- **Distribution-aware Imputation**:
  - For skewed variables `Water_1000kcal` and `SW_Water_1000kcal`, missing values were imputed using:
    - **Mean** if the feature was approximately normally distributed (`-1 < skew < 1`).
    - **Median** if the feature was highly skewed.

- **Post-imputation Check**:
  - Confirmed that all selected features were now complete (no missing values).
  - Updated `df_aqinfo` to reflect the final status of each feature after imputation.

- **Final Dataset Structure**:
  - Re-validated the dataset structure and memory usage using `df_aq.info()`.
  - Ensured all transformations and imputations were appropriately applied.

```python
# Define the order of imputation
imputation_order = ["Land_100gP", "Eutro_100gP", "GHG_100gP", "Water_100gP", "SW_Water_100gP"]

# Define the predictors for each target
predictors = {
    "Land_100gP": ["Farm", "Total_emissions", "GHG_100gP", "Land_1000kcal", "Land_kg"],
    "Eutro_100gP": ["Eutro_kg", "Eutro_1000kcal", "GHG_100gP", "Land_100gP"],
    "GHG_100gP": ["Eutro_100gP", "Land_100gP"],
    "Water_100gP": ["SW_Water_100gP"],
    "SW_Water_100gP": ["Water_100gP"]
}

def regression_impute(df, target, predictors):
    df_copy = df.copy()
    
    # Identify missing rows
    missing_rows = df_copy[df_copy[target].isna()]
    if missing_rows.shape[0] == 0:
        print(f"No missing values for {target}, skipping...")
        return df_copy
    
    # Training data (drop rows where target is missing)
    train_df = df_copy.dropna(subset=[target])  
    X_train = train_df[predictors]
    y_train = train_df[target]
    
    # Ensure predictors have no missing values in the training data only
    imp = SimpleImputer(strategy="median")
    X_train = imp.fit_transform(X_train)
    
    # Train regression model
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    # Prepare missing data for prediction
    X_missing = df_copy.loc[df_copy[target].isna(), predictors]
    
    # Only fill missing predictor values for regression input (don't overwrite df globally)
    X_missing = imp.transform(X_missing)
    
    # Predict missing values
    df_copy.loc[df_copy[target].isna(), target] = model.predict(X_missing)
    
    print(f"Missing values in {target} filled using regression.")
    return df_copy

# Apply regression imputation in the correct order
for target in imputation_order:
    df_aq = regression_impute(df_aq, target, predictors[target])

missing_counts = df_aq.isna().sum()
non_null_counts = df_aq.notna().sum()
dtype_info = df_aq.dtypes

df_aqinfo = pd.DataFrame({
    "Non-Null Count": non_null_counts,
    "Missing Count": missing_counts,
    "Missing Percentage": round((missing_counts / len(df_aq)) * 100, 2),
    "Dtype": dtype_info
})

# Filter and display only the info for the selected columns  
df_selected_info = df_aqinfo.loc[["Water_100gP", "SW_Water_100gP", "Land_100gP", "Eutro_100gP", "GHG_100gP", "Water_1000kcal", "SW_Water_1000kcal"]]  
df_selected_info
```

```python
for col in ['Water_1000kcal', 'SW_Water_1000kcal']:
    skewness = df_aq[col].skew()
    
    if -1 < skewness < 1:  # Approximately normal distribution
        df_aq[col].fillna(df_aq[col].mean(), inplace=True)
    else:  # Highly skewed (either positive or negative)
        df_aq[col].fillna(df_aq[col].median(), inplace=True)

missing_counts = df_aq.isna().sum()
non_null_counts = df_aq.notna().sum()
dtype_info = df_aq.dtypes

df_aqinfo = pd.DataFrame({
    "Non-Null Count": non_null_counts,
    "Missing Count": missing_counts,
    "Missing Percentage": round((missing_counts / len(df_aq)) * 100, 2),
    "Dtype": dtype_info
})

# Filter and display only the info for the selected columns  
df_selected_info = df_aqinfo.loc[["Water_100gP", "SW_Water_100gP", "Land_100gP", "Eutro_100gP", "GHG_100gP", "Water_1000kcal", "SW_Water_1000kcal"]]  
df_selected_info
```

##### <a id='toc1_2_3_7_2_'></a>[2.3.7.2. Statistical Testing Approach: Parametric and Non-Parametric Tests ](#toc0_)

```python
# Define custom color mapping
color_map = {
    "Reject H₀: Animal-based foods have a significantly higher impact.": "#A52A2A",  # Grey-Red 
    "Fail to reject H₀: No significant difference.": "#4682B4"   # Grey-Blue
}

# Select relevant metrics and rename for better representation
metric_names = {
    "Total_emissions": "GHG Emissions_kg",
    "Land_kg": "Land Use_kg",
    "Water_kg": "Water Consumption_kg"
}

# Initialize results dictionary
results = {}

# Perform tests for each metric
for metric, display_name in metric_names.items():
    # Separate animal-based and plant-based values
    animal_values = df_aq.loc[df_aq["Food_Category"] == "Animal-Based Foods", metric]
    plant_values = df_aq.loc[df_aq["Food_Category"] == "Plant-Based Foods", metric]

    # Perform independent t-test
    t_stat, p_val_ttest = stats.ttest_ind(animal_values, plant_values, equal_var=False)

    # Interpretation for t-test
    alpha = 0.05
    interpretation_ttest = (
        "Reject H₀: Animal-based foods have a significantly higher impact."
        if p_val_ttest < alpha else
        "Fail to reject H₀: No significant difference."
    )

    # Perform Mann-Whitney U-Test
    u_stat, p_val_uw = stats.mannwhitneyu(animal_values, plant_values, alternative='two-sided')

    # Interpretation for Mann-Whitney U-Test
    interpretation_uw = (
        "Reject H₀: Animal-based foods have a significantly higher impact."
        if p_val_uw < alpha else
        "Fail to reject H₀: No significant difference."
    )

    # Store results for both tests
    results[display_name] = {
        "t-statistic": t_stat, "p-value (T-Test)": p_val_ttest, "Interpretation (T-Test)": interpretation_ttest,
        "U-statistic": u_stat, "p-value (U-Test)": p_val_uw, "Interpretation (U-Test)": interpretation_uw
    }

# Convert results to DataFrame
df_results = pd.DataFrame.from_dict(results, orient="index").reset_index()
df_results.rename(columns={"index": "Metric"}, inplace=True)

# Custom hover template for T-Test
hover_template_ttest = (
    "<b>Metric:</b> %{x}<br>"
    "<b>T-Statistic:</b> %{y:.3f}<br>"
    "<b>P-Value:</b> %{customdata[0]:.4f}<br>"
    "<b>Decision:</b> %{customdata[1]}<extra></extra>"
)

# Custom hover template for Mann-Whitney U-Test
hover_template_uw = (
    "<b>Metric:</b> %{x}<br>"
    "<b>U-Statistic:</b> %{y:.3f}<br>"
    "<b>P-Value:</b> %{customdata[0]:.4f}<br>"
    "<b>Decision:</b> %{customdata[1]}<extra></extra>"
)

# Create a bar chart for T-test statistics
fig_ttest = px.bar(
    df_results,
    x="Metric",
    y="t-statistic",
    color="Interpretation (T-Test)",
    title="<b>T-Test Results: Environmental Impact of Animal vs. Plant-Based Foods</b>",
    labels={"t-statistic": "T-Statistic", "Metric": "Environmental Metric"},
    color_discrete_map=color_map,  # Apply custom colors
    custom_data=["p-value (T-Test)", "Interpretation (T-Test)"]
)

# Apply hover template for T-test
fig_ttest.update_traces(
    hovertemplate=hover_template_ttest,
    texttemplate="p = %{customdata[0]:.4f}",
    textposition="outside"
)

# Create a bar chart for Mann-Whitney U-test statistics
fig_uw = px.bar(
    df_results,
    x="Metric",
    y="U-statistic",
    color="Interpretation (U-Test)",
    title="<b>Mann-Whitney U-Test Results: Environmental Impact of Animal vs. Plant-Based Foods</b>",
    labels={"U-statistic": "U-Statistic", "Metric": "Environmental Metric"},
    color_discrete_map=color_map,  # Apply custom colors
    custom_data=["p-value (U-Test)", "Interpretation (U-Test)"]
)

# Apply hover template for Mann-Whitney U-test
fig_uw.update_traces(
    hovertemplate=hover_template_uw,
    texttemplate="p = %{customdata[0]:.4f}",
    textposition="outside"
)

# Layout adjustments for T-Test plot
fig_ttest.update_layout(
    title_x=0.5,  # Center title
    yaxis_title="T-Statistic",
    xaxis_title="Environmental Metric",
    height=600, width=1200
)

# Layout adjustments for U-Test plot
fig_uw.update_layout(
    title_x=0.5,  # Center title
    yaxis_title="U-Statistic",
    xaxis_title="Environmental Metric",
    height=600, width=1200
)

# Show figures
fig_ttest.show()
fig_uw.show()

# Show the final results DataFrame
df_results
```

`Comment` 

The decision between a **t-test** and the **Mann-Whitney U (MWU) test** depends on the assumptions of **normality** and **homogeneity of variance**:  

1. **Non-Normal Distributions**:  
   - The skewness values for **GHG emissions (1.56 vs. 2.34)**, **land use (1.25 vs. 2.99)**, and **water consumption (1.05 vs. 2.22)** indicate strong right-skewed distributions, particularly for plant-based foods.  
   - Kurtosis values for plant-based foods (>3) suggest **heavy-tailed distributions**, further deviating from normality.  

2. **Heterogeneity of Variance**:  
   - The **GHG variance (327.47 vs. 23.39)**, **land use variance (20,911.84 vs. 217.90)**, and **water variance (2,794,369 vs. 982,888)** are significantly different between groups.  
   - This violates the equal variance assumption required for a **t-test** (even Welch’s t-test may be unreliable with extreme variance differences).

Given the **non-normality** and **heterogeneous variances** observed in the dataset, the **Mann-Whitney U (MWU) test** remains the more appropriate statistical test. It is a **non-parametric** test that does not assume normal distributions or equal variances, making it more **robust** for this dataset. This is further supported by the test results:  

1. **Inconsistencies in T-Test Results**:  
   - The **t-test fails to reject H₀ for GHG emissions (p = 0.054) and land use (p = 0.103)**, meaning it does not detect a significant difference between food categories.  
   - However, it **rejects H₀ for water consumption (p = 0.034)**, showing an inconsistent interpretation across metrics.  

2. **MWU Test Shows Stronger Evidence**:  
   - The **MWU test rejects H₀ for all metrics (p < 0.005 for all cases)**, consistently indicating that **animal-based foods have a significantly higher impact** across GHG emissions, land use, and water consumption.  
   - This aligns with the **data’s skewed distributions and unequal variances**, where MWU provides more reliable conclusions.  


**Conclusion**  
Based on the **Mann-Whitney U test results**, we **reject the null hypothesis (H₀)** across **all three environmental impact metrics** (GHG emissions, land use, and water consumption) since **p-values are all below 0.05**.
  
Thus, we **accept the alternative hypothesis (H₁):**  
*"On average, animal-based foods have a significantly higher environmental impact across GHG emissions, land use, and water consumption compared to plant-based foods."*
